### You are creating a student result calculator.

A student has:

    Mathematics: 85
    English: 72
    Python: 90
    Total possible marks per subject: 100
    Student name: "Ahmed"
    Required passing marks per subject: 50

### Create variables and:

    Calculate the student's total marks.
    Calculate the average marks.
    Calculate the percentage.
    Check whether the student passed Mathematics.
    Check whether the student passed all three subjects.
    Check whether the name contains the letter "A".
    Check whether the average marks are greater than or equal to 80.
    Store the final percentage and check whether it is truthy.

Print all results.

```python
math = 85
english = 72
python = 90
total_possible = 100
student_name = "Ahmed"
passing_marks = 50

# Total marks
total_marks = math + english + python

# Average marks
average = total_marks / 3

# Percentage
percentage = (total_marks / (total_possible * 3)) * 100

# Passed Mathematics
passed_math = math >= passing_marks

# Passed all three subjects
passed_all = math >= passing_marks and english >= passing_marks and python >= passing_marks

# Name contains the letter "A"
name_check = "A" in student_name

# Average marks >= 80
average_check = average >= 80

# Store final percentage and check whether it is truthy
final_percentage = percentage
percentage_truthy = bool(final_percentage)

# Print all results
print("Total marks:", total_marks)
print("Average marks:",average)
print("percentage:",percentage)
print("passed mathematics:",passed_math)
print("passed all subject:",passed_all)
print('name contains"A":',name_check)
print("average>=80:",average_check)
print("percentage is truthy:",percentage_truthy)
```
